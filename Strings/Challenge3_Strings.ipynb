## String-Based Questions

### Count frequency of characters

```python
inp_str=input("Enter the string : ")
print(f"----------Frequency of Characters-------------\nInput String : {inp_str}")
for char in inp_str:
  print(f"Frequency of '{char}' = {inp_str.count(char)}")
```

### Find first non-repeating character

```python
# first non-repeating character
inp_str=input("Enter the String : ")
for char in inp_str:
  if inp_str.count(char)==1:
    print(f"First non-repeating Character = {char}")
    break
```

### Remove spaces from string

```python
# Remove spaces from string
inp_str=input("Enter the String : ")
final_str=""
for item in inp_str.split():
  final_str+=item
print(f"String '{inp_str}' spaces removed : {final_str}")
```

### Convert string to uppercase/lowercase

```python
# Convert string to uppercase/lowercase
inp_str=input("Enter the String : ")
print(f"String '{inp_str}' in lowercase : {inp_str.lower()}")
print(f"String '{inp_str}' in uppercase : {inp_str.upper()}")
```

### Check anagram

```python
# Check anagram
print("-----------Check anagram----------------------------")
inp_str1=input("Enter the String1 : ").replace(" ","").lower()
inp_str2=input("Enter the String2 : ").replace(" ","").lower()
if sorted(inp_str1)==sorted(inp_str2):
  print(f"{inp_str1} and {inp_str2} are anagram")
else:
  print(f"{inp_str1} and {inp_str2} are not an anagram")
```

### Replace vowels with *

```python
# Replace vowels with *
inp_str=input("Enter the String : ")
copy_str=inp_str
for char in inp_str:
  if char in ['a','e','i','o','u']:
    inp_str=inp_str.replace(char,'*')
print(f"Vowels replaced in '{copy_str}' = {inp_str}")
```

### Count words in a sentence

```python
# Count words in a sentence

inp_str=input("Enter the Sentence : ")
print(f"No of words in the sentence = {len(inp_str.split())}")
```

### Find longest word in a sentence

```python
# Find longest word in a sentence
inp_str=input("Enter the String : ")
inp_list=inp_str.split()
length=1
longest_word=""
for item in inp_list:
  if len(item)>length:
    length=len(item)
    longest_word=item
print(f"Longest word in the sentence = {longest_word} and no. of chars = {length}")
```

### Reverse each word in a sentence

```python
# Reverse each word in a sentence
inp_str=input("Enter the String : ")
reversed_list=[item[::-1] for item in inp_str.split()]
reversed_str=""
for item in reversed_list:
  reversed_str+=item+" "
print(f"Original String : {inp_str}\nReversed each word in the string : {reversed_str}")
```

### Remove duplicate characters

```python
# Remove duplicate characters
inp_str=input("Enter the Sentence : ")
no_dupli_str=""
for char in inp_str:
  if char not in no_dupli_str:
    no_dupli_str+=char
print(f"Duplicate Characters Removed from string '{inp_str}' = {no_dupli_str}")
```
