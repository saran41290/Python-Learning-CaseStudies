## Beginner Questions


### 1.Print first and last character of a string.


```python
inp_str=input("Enter your string : ")
print(f"The first char in the string '{inp_str}' = {inp_str[0]}\nThe last char in the string '{inp_str}' = {inp_str[-1]}")
```

### 2.Reverse a string using slicing.


```python
inp_str=input("Enter your string : ")
print(f"Reversed string '{inp_str}' = {inp_str[::-1]}")
```

### 3.Count vowels in a string.


```python
inp_str=input("Enter your string : ")
count=0
for letter in inp_str:
    if letter in ['a','e','i','o','u']:
      count+=1
print(f"The number of vowels in '{inp_str}' = {count}")
```

### 4.Convert string to uppercase and lowercase.


```python
inp_str=input("Enter your string : ")
print(f"Input String : {inp_str}\nUppercase String : {inp_str.upper()}\nLowercase String : {inp_str.lower()}")
```

### 5.Replace spaces with _.


```python
inp_str=input("Enter your string : ")
print(f"Replaced ' ' with '_' : {inp_str.replace(' ','_')}")
```

## Intermediate Questions


### 6.Check whether a string is palindrome or not.


```python
inp_str=input("Enter your string : ")
if(inp_str == inp_str[::-1]):
  print(f"The String '{inp_str}' is a palindrome!")
else :
  print(f"The String '{inp_str}' is not a palindrome!")
```

### 7.Count occurrence of each character in a string.


```python
inp_str=input("Enter your string : ")
for letter in inp_str:
  print(f"Count of '{letter}' in '{inp_str}' = {inp_str.count(letter)}")
```

### 8.Find the longest word in a sentence.


```python
inp_str=input("Enter your sentence : ")
inp_list=inp_str.split()
len_list=[]
for word in inp_list:
  len_list.append(len(word))
print(len_list)
index=len_list.index(max(len_list))
print(f"The Longest word in the sentence '{inp_str}' is '{inp_list[index]}'")

```

### 9.Remove duplicate characters from a string.


```python
inp_str=input("Enter your string : ")
res_str=""
for letter in inp_str:
  if letter not in res_str:
    res_str+=letter

print(f"Duplicate Characters removed from string '{inp_str}' = '{res_str}'")
```

### 10.Extract domain name from email.

## Advanced / Data Analyst Style


### 1.Clean customer names dataset:
["  saran", "ARUN ", "viJAY"]
Expected:
["Saran", "Arun", "Vijay"]

```python
inp_list=[" saran", "ARUN ", "viJAY"]
output_list=[]
for word in inp_list :
  output_list.append(word.title())
print(f"Cleaned Customer Names List : {output_list}")
```

### 2.Extract hashtags from social media text.

```python
import re
inp_str=input("Enter the social media text to extract hashtags : ")
hash_list=[word for word in inp_str.split() if word.startswith('#')]
hashtags=re.findall(r"#\w+",inp_str)
print(f"extracted HashTags are : {hash_list}\n{hashtags}")
```

### 3.Find top occurring word in paragraph.

```python
inp_str=input("Enter your paragraph : ")
inp_list=inp_str.split()
count_list=[]
for word in inp_str:
  count_list.append(inp_str.count(word))
index=max(count_list)
print(f"Top Occuring Word in the paragraph : {inp_list[index]}")
```

### 5.Create username generator from full name.
