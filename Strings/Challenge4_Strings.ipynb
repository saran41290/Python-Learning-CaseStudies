## String

### Find the length of a string

```python
# Find the length of a string
inp_str=input("Enter the string : ")
print(f"Length of '{inp_str}' = {len(inp_str)}")
```

### Convert a string to uppercase

```python
inp_str=input("Enter the string : ")
print(f"Uppercase of '{inp_str}' = {inp_str.upper()}")
```

### Convert a string to lowercase

```python
inp_str=input("Enter the string : ")
print(f"Lowercase of '{inp_str}' = {inp_str.lower()}")
```

### Check if a substring exists in a string

```python
# Check if a substring exists in a string
inp_str=input("Enter the string : ")
substr=input("Enter the sub-string to find in a string : ")
count=inp_str.count(substr)
if(count>1):
  print(f"sub-string '{substr}' is present in the string\nNo.of times the sub-string '{substr}' found = {count} ")
else:
  print(f"sub-string '{substr}' is not present in the string")
```

### Count how many times a character appears in a string

```python
inp_str=input("Enter the string : ")
char=input("Enter the character to find : ")
print(f"No. of times '{char}' present in string = {inp_str.count(char)}")
```

### Replace a word in a string

```python
inp_str=input("Enter the string : ")
find=input("Enter the word to replace : ")
replac=input("Enter the word to be replaced : ")
print(f"Replaced String :  {inp_str.replace(find,replac)}")
```

### Remove leading and trailing spaces

```python
inp_str=input("Enter the string : ")
print(f"Leading and trailing spaces removed : '{inp_str.strip()}'")
```

### Split a string into a list Join a list into a string


```python
inp_str=input("Enter the string : ")
inp_list=inp_str.split()
print(f"String to List : {inp_list}")
final_str=""
for item in inp_list:
  final_str+=item+" "
print(f"List back to String : {final_str}")
```

### Print the first character,the last character of a string

```python
inp_str=input("Enter the string : ")
print(f"First char in string : {inp_str[0]}")
print(f"First char in string : {inp_str[-1]}")
```

### Extract a substring using slicing

```python
#Extract a substring using slicing
inp_str=input("Enter the string '(2026-05-19)': ")
print(f"Extrected Sub-strings:\nYear = {inp_str[0:4]}\nMonth = {inp_str[5:7]}\nDay = {inp_str[8:]}")


```

### Reverse a string using slicing

```python
inp_str=input("Enter the string : ")
print(f"Reversed String : {inp_str[::-1]}")
```

### Check if a string is a palindrome

```python
inp_str=input("Enter the string : ")
if inp_str[::-1]==inp_str:
  print(f"'{inp_str}' is a palindrome")
else:
  print(f"'{inp_str}' is not a palindrome")
```

### Count the number of consonants in a string

```python
inp_str=input("Enter the string : ")
count=0
for char in inp_str:
  if char not in ['a','e','i','o','u']:
    count+=1
print(f"No. of consonants in string = {count}")
```

### Find the number of words in a sentence

```python
inp_str=input("Enter the string : ")
print(f"No. of words in the sentence = {len(inp_str.split())}")
```

### Convert the first letter of each word to uppercase

```python
inp_str=input("Enter the string : ")
print(f"Convert the first letter of each word to uppercase:\nOriginal String = {inp_str}")
final_str=""
for item in inp_str.split():
  final_str+=item.title()+" "
print(f"Final String : {final_str}")
```
