## Task1 : Student Registration System

## Task1 : Email Username Generator

```python
'''
Case Study – Email Username Generator
Instructions:

• Create variables: full_name, domain_name

• Take input from the user using input() function

• Convert full_name into lowercase using lower()

• Remove spaces in name and replace with dot using replace()

• Split full_name using split() function

• Join name parts using join() function if needed

• Create email in format: name@domain

• Display only Full Name and Generated Email

• Use print() and f-string for output display
'''
full_name=input("Enter your full name : ")
domain_name=input("Enter your domain_name : ")
full_name=full_name.lower().strip().replace(" ","_")
splitted_name=full_name.split("_")
full_name="_".join(splitted_name)
print(f"Splitted name : {splitted_name}\nJoined Full name : {full_name}")
email=full_name+'@'+domain_name
print(f"\n\n------Email UserName Generator---------\nFull name : {full_name}\nGenerated EmailUserName : {email}")
```

## Task2 : Student ID Card Generator

```python
"""
Case Study – Student ID Card Generator
Instructions:

• Create variables: student_name, roll_number, department

• Take input from the user using input() function

• Convert student_name and department into uppercase using upper()

• Extract first 4 letters of student_name and department using slicing

• Create Student ID by combining name, department, and roll number

• Display Name, Department, Roll Number, and Student ID only

• Use print() and f-string for output display
"""
stu_name=input("Enter the student name : ").upper().strip()
roll_num=int(input("Enter the student roll number : ").strip())
dep=input("Enter the department : ").upper().strip()
student_id=dep[0:4]+stu_name[0:4]+str(roll_num)
print(f"\n\n--------------Student ID Card Generator---------------\nGenerated StudentIDCard : {student_id}")

```

## Task1 : Student Details using Tuple (Packing & Unpacking)
Instructions:

Create a tuple named student and store the following details: name, age, course

Unpack the tuple into separate variables: name, age, course

Display each value separately using print().

```python
#Create a tuple named student and store the following details: name, age, course
student=('Saranya',34,'DataAnalytics')
print("----------Unpacking Tuple----------")
name,age,course=student
print(f"Student Details:\nName:{name}\nAge:{age}\nCourse:{course}")
print("--------Packing Tuple----------")
student=name,35,course
print(f"Student Details:\n{student}")
```

##Task1 : Shopping Cart System
Instructions:

Create a list named cart with initial items: Milk, Bread, Eggs, Apple

Perform list operations:

• Add multiple items using extend()

• Insert item at index 2 using insert()

• Remove item Apple using remove()

• Remove last item using pop()

Use list methods:

• index() to find position of Eggs

• count() to count occurrences of Milk

Apply indexing:

• Display first item

• Display last item

Apply slicing:

• Display first 3 items from the list

Find total number of items using len()

Display all outputs using print()

```python
cart=['Milk','Bread','Eggs','Apple']
cart.extend(['Banana','Omelette','Fresh-Juice'])
print(cart)
#Insert item at index 2 using insert()
cart.insert(2,'Coffee')
print(f"Insert item at index 2 using insert():\n{cart}")
cart.remove('Apple')
print(f" Remove item Apple using remove() :\n{cart}")
cart.pop()
print(f" Remove item using pop() :\n{cart}")
#Apply indexing:• Display first item• Display last item
print(f"\n------------------Indexing--------------------")
print(f"First item in cart : {cart[0]}")
print(f"Last item in cart : {cart[-1]}")
#Apply slicing:• Display first 3 items from the list
print(f"\n---------------Slicing------------------")
print(f"First 3 items in list : {cart[0:3]}")
#Find total number of items using len()
print(f"\n\nTotal number of items : {len(cart)}")
#Display all outputs using print()
print(f"Final cart : {cart}")
```

##Task2 : Student Marks Analysis
Instructions:

Create a list named marks by getting 5 subject marks from the user using map() and list()

Perform calculations:

• Find total marks using sum()

• Find average marks using len()

• Calculate percentage assuming each subject is out of 100

```python
print(f"----------------Student Marks Analysis-----------------")
inp_marks=list(map(int,input("Enter the subject marks separated by ',': ").split(',')))
print(f"Input marks : {inp_marks}")

print(f"1.Total Marks={sum(inp_marks)}")
print(f"2.Average Marks={round(sum(inp_marks)/len(inp_marks),2)}")
#Percentage=(Total Maximum Marks/Total Marks Obtained)*100
#Total Maximum Marks = Number of Subjects × 100
avg_marks=sum(inp_marks)/len(inp_marks)
percentage = round(avg_marks/100,2)
print(f"3.Percentage={percentage}")
print(f"4.Maximum Marks={max(inp_marks)}")
print(f"5.Minimum Marks={min(inp_marks)}")
```
