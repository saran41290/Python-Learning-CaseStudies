#Set Questions


##Beginner


###1.Remove duplicates using set.


```python
list1=[100,101,120,100,101,156,178,199,199,200]
set1=set(list1)
print(f"Initial List = {list1}\nSet Created = {set1}")

```

###2.Find union and intersection.


```python
a={100,200,101,204,209,120}
b={200,210,220,209}
print(f"set1 = {a}\nset2 = {b}\n1.Union = {a.union(b)}\n2.Intersection = {a.intersection(b)}")
```

###3.Check subset and superset.


```python
a={100,200,101,204,209,120}
b={200,209}
print(f"set1 = {a}\nset2 = {b}\n1.A Subset of B = {a.issubset(b)}\n2.B Subset of A = {b.issubset(a)}")
print(f"3.A Superset of B = {a.issuperset(b)}\n4.B Superset of A = {b.issuperset(a)}")
```

##Intermediate


###1.Find common skills between two employees.


```python
emp1={"c","c++","java","python","c#","SQL"}
emp2={"SQL","python","java","HTML","PHP"}
print(f"Employee 1 Skills = {emp1}\nEmployee 2 Skills = {emp2}\nCommon Skills ={emp1.intersection(emp2)}")
```

###1.Market basket analysis:
Customers purchased:
A = {"Milk", "Bread", "Butter"}
B = {"Bread", "Eggs", "Butter"}
Find:
common products
recommended products

```python
A = {"Milk", "Bread", "Butter"}
B = {"Bread", "Eggs", "Butter"}
print(f"A PurchaseList = {A}\nB PurchaseList = {B}")
print(f"Recommended Products = {A.intersection(B)}")
```
