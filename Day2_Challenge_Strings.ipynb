# Day 2 Challenge - Strings

## Question 1- perform all arithmetic operations

```python
'''
Day 2:
1.Read two numbers and perform all arithmetic operations
2.Check the data type of the value entered
3.Find the length of a string and count the number of words
4.Count words in a sentence
5.Reverse each word in a sentence
6.Check whether a character is a vowel or consonant
7.Check if a string is a palindrome
'''
print("*************Question 1- perform all arithmetic operations**********************\n")
num1=float(input("Enter the first number : ").strip())
num2=float(input("Enter the second number : ").strip())
print("\n-------------Arithmetic Opeartions--------------")
print(f"1.Addition of 2 numbers({num1},{num2}) = {(num1+num2):.2f}")
print(f"2.Subtraction of 2 numbers({num1},{num2}) = {(num1-num2):.2f}")
print(f"3.Multiplication of 2 numbers({num1},{num2}) = {(num1*num2):.2f}")
print(f"4.Division of 2 numbers({num1},{num2}) = {(num1/num2):.2f}")
```

## Question 2 - Check the data type of the value entered

```python
print("\n*************Question 2 - Check the data type of the value entered**********************\n")
print(f"The datatype of the entered value : {type(input("Enter the value : "))}")
```

## Question 3 - Find the length of a string and count the number of words

```python
print("\n*************Question 3 - Find the length of a string and count the number of words**********************\n")
inp_str=input("Enter the sentence : ").strip()
print(f"The Length of the String = {len(inp_str)}\nNumber of words in the string = {len(inp_str.split())}")
```

## Question 4 - Count words in a sentence

```python
print("\n*************Question 4 - Count words in a sentence**********************\n")
inp_str=input("Enter the sentence : ").strip()
print(f"Number of words in the string = {len(inp_str.split())}")
```

## Question 5 - Reverse each word in a sentence

```python
print("\n*************Question 5 - Reverse each word in a sentence**********************\n")
inp_str=input("Enter Your Sentence : ")
inp_str_list=inp_str.split()
rev_list=[]
for word in inp_str_list:
  rev_list.append(word[::-1])
print(f"Entered Sentence : {inp_str}\nEach word reversed in the sentence : {" ".join(rev_list)}")
```

## Question 6 - Check whether a character is a vowel or consonant

```python
print("\n*************Question 6 - Check whether a character is a vowel or consonant**********************\n")
inp_char=input("Enter the character ").lower().strip()
vowel_list=['a','e','i','o','u']
if inp_char in vowel_list :
  print(f"Character {inp_char} is a Vowel")
else :
  print(f"Character {inp_char} is a Consonant")
```

## Question 7 - Check if a string is a palindrome

```python
print("\n*************Question 7 - Check if a string is a palindrome**********************\n")
inp_str=input("Enter your String to check :").strip().lower()
if(inp_str == inp_str[::-1]):
  print(f"The String {inp_str} is a Palindrome")
else :
  print(f"The String {inp_str} is not a Palindrome")
```
