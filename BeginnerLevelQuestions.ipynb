### Check if a number is even or odd

```python
print("----------Odd / Even Number-----------------")
inp_num=int(input("Enter your number : ").strip())
if inp_num % 2 == 0:
  print(f"{inp_num} is Even")
else :
  print(f"{inp_num} is Odd")
```

### Find largest of 3 numbers

```python
print("-------------Find largest of 3 numbers-----------------")
inp_str=input("Enter the 3 numbers separated by , : ")
inp_list=inp_str.split(',')
print(f"Maximum of 3 numbers : { max(inp_list) }")
```

### Check if a number is prime

```python
num = int(input("Enter a number: "))

if num <= 1:
    print("Not Prime")

else:

    is_prime = True

    for i in range(2, int(num**0.5) + 1):

        if num % i == 0:
            is_prime = False
            break

    if is_prime:
        print(f"{num} is Prime")

    else:
        print(f"{num} is Not Prime")
```

### Print factorial of a number

```python
inp_num=int(input("Enter the number : "))
factorial=1
for i in range(1,inp_num+1):
  factorial*=i
print(f"Factorial of {inp_num} = {factorial}")
```

### Reverse a string

### Check palindrome string

```python
print("-----------------Check palindrome string------------")
inp_str=input("Enter the string to check : ")
if(inp_str==inp_str[::-1]):
    print(f"{inp_str} is a palindrome")
else :
    print(f"{inp_str} is not a palindrome")
```

### Count vowels in a string

```python
inp_str=input("Enter your string : ")
count=0
for char in inp_str:
  if char in ['a','e','i','o','u']:
    count+=1
print(f"Number of Vowels in '{inp_str}' = {count}")
```

### Find sum of digits of a number

```python
num=input("Enter the  number to find the sum of digits : ")
num_list=list(map(int,num))
print(f"Sum of digits in number '{num}' = {sum(num_list)}")
```
