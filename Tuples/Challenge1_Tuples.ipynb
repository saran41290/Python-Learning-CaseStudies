#Beginner Questions


##1.Create tuple of 5 fruits.
Access:
first element
last element
Count occurrence of a value.
Find index of a value.



```python
fruits=('apple','mango','apple','jack-fruit','dragon-fruit','pineapple','guava','guava')
print(f"1st Element in tuple : {fruits[0]}")
print(f"Last Element in tuple : {fruits[-1]}")
print(f"Occurrences of 'apple'  in tuple : {fruits.count("apple")}")
print(f"Index of 'dragon-fruit' in tuple : {fruits.index("dragon-fruit")}")
```

#Intermediate Questions

##1.Convert tuple to list and modify it.


```python
print("Converting Tuple to List")
fruits_list=list(fruits)
print(f"List of fruits : {fruits_list}")
fruits_list[0]='strawberry'
print(f"List of fruits after modify : {fruits_list}")
```

##2.Swap two tuples.


```python
tuple1=(10,20,30,40)
tuple2=(1,2,3,4)
print(f"Before Swapping:\nTuple1:{tuple1}\nTuple2:{tuple2}")
tuple1,tuple2=tuple2,tuple1
print(f"Before Swapping:\nTuple1:{tuple1}\nTuple2:{tuple2}")
```

##3.Find max and min element.


```python
print(f"Tuple Values :{tuple2}")
print(f"Maximum Element:{max(tuple2)}")
print(f"Minimum Element:{min(tuple2)}")
```

##4.Unpack tuple values into variables.



```python
print(f"Unpacking Tuples into variables\nTuple Values:{tuple2}")
a,b,c,d=tuple2
print(f"a={a},b={b},c={c},d={d}")
```

#Advanced Questions


##1.Store employee records using tuples:
(id, name, salary)


```python
emp_rec=((111,'saran',1000.89),(102,'Arasu',8900.98),(99,'Mithra',1200.98),(104,'Sharvesh',7099.67))
print(f"--------------Store employee records using tuples-------------------")
print(f"Employee Details: {emp_rec}")
```

##2.Sort tuple data based on salary.


```python
emp_rec_sorted=sorted(emp_rec,key=lambda emp_rec:emp_rec[2],reverse=True)
print(f"Sorted employee tuple:{emp_rec_sorted}")
```

##3.Create nested tuple and access inner values.

```python
print(f"Nested Tuple Employee Records:{emp_rec}")
print(f"Nested tuple value at pos[0][2]={emp_rec[0][2]}")
```
