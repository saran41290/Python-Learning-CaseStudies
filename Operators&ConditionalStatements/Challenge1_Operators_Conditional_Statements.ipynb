###2.ATM withdrawal simulation.


```python
balance = 10000

amount = int(input("Enter withdrawal amount: "))

if amount <= 0:
    print("Invalid amount")

elif amount > balance:
    print("Insufficient balance")

else:
    balance = balance - amount
    print(f"Withdrawal successful!\nAmount withdrawn = {amount}")
    print(f"Remaining balance = {balance}")
```

###4.Login authentication system.



```python
#Create variables: • password • attempts • max_attempts
password='my12345'
attempts=0
max_attempts=3
#Use a while loop to allow the user to enter the password.
while(attempts < max_attempts):
#Check whether the entered password matches the original password.
  inp_pwd=input("\nEnter your password : ")
#If the password is correct: • Display "Access granted!" • Stop the loop using break
  if inp_pwd == password :
    print(f"\nLogin Successful!")
    break
#If the password is incorrect: • Increase attempt count • Display remaining attempts
  else:
    attempts+=1
    print(f"Incorrect Password!\nNo. of attempts remaining = {max_attempts-attempts}")

else:
  print(f"Maximum attempts reached. Access denied!")
#If maximum attempts are reached: • Display "Maximum attempts reached. Access denied!"
```

###1.Income tax calculator.
Income up to ₹2,50,000 → 0%
₹2,50,001 – ₹5,00,000 → 5%
₹5,00,001 – ₹10,00,000 → 20%
Above ₹10,00,000 → 30%


```python
income = float(input("Enter annual income: "))

if income <= 250000:
    tax = 0

elif income <= 500000:
    tax = income * 0.05

elif income <= 1000000:
    tax = income * 0.20

else:
    tax = income * 0.30

net_income = income - tax

print("Tax to be paid = ", tax)
print("Income after tax deducted = ", net_income)
```

###2.E-commerce discount engine:
Conditions:
Premium customer → 10% discount
Festive discount → 15% discount
Coupon discount → ₹500 off


```python
amount = float(input("Enter purchase amount: "))

premium = input("Premium customer? (yes/no): ")
festive = input("Festive offer? (yes/no): ")
coupon = input("Coupon available? (yes/no): ")

discount = 0

# premium discount
if premium == "yes":
    discount += amount * 0.10

# festive discount
if festive == "yes":
    discount += amount * 0.15

# coupon discount
if coupon == "yes":
    discount += 500

final_amount = amount - discount

# avoid negative bill
if final_amount < 0:
    final_amount = 0

print("Total discount = ", discount)
print("Final bill = ", final_amount)
```

###3.Loan eligibility checker.


```python
#Create variables: customer_name, age, income, credit_score, existing_loan
customer_name, age, income, credit_score, existing_loan='saranya',35,40000,600,'no'
status=""
print(f"--------------Bank Loan Eligibility System----------------")

#Use conditional statements (if, elif, else) to:
#• Approve loan if age ≥ 21, income ≥ 30000, credit_score ≥ 700, and no existing loan
#• Conditionally approve loan if credit_score ≥ 600 and income ≥ 40000
#• Otherwise reject the loan application
if age >= 21 and income >= 30000:
  if credit_score >= 700 and existing_loan=='no':
        status = "Approved "
  elif credit_score >= 600 and income >= 40000:
        status = "Conditionally Approved "
  else:
        status = "Rejected "
else :
  status = "Not Eligible"
#Display customer details and loan status using print()
print(f"Customer details :\nName : {customer_name}\nAge : {age}\nIncome : {income}\nCredit Score : {credit_score}\nExisting Loan(if any) : {existing_loan}")
print(f"Status of your Loan Application : {status}")
```
