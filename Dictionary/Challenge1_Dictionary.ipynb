#Dictionary



## Beginner


###1.Create dictionary of student marks.


```python
student_marks={
    "saranya":100,
    "mithra":98,
    "sai":87,
    "sharvesh":78,
    "arasu":90
}
student_marks
```

```python
print(f"-----Original Dictionary----------\n{student_marks}")
student_marks["kavya"]=99
student_marks
```

##Intermediate


###1.Merge two dictionaries.


```python
dict1={"101":"laptop","102":"mouse","103":"key-board","104":"printer"
      }
dict2={"laptop":10000,
       "mouse":800,
       "key-board":900,
       "printer":1100
      }
merged_dict={**dict1,**dict2}
print(f"Original Dictionaries\nDict1 : {dict1}\nDict2 : {dict2}")
dict1.update(dict2)
print(f"Merged Dictionaries\n1.using '|' operator : {dict1|dict2}\n2.Using update : {dict1}\n3.Using ** Unpacking : {merged_dict}")
```

###2.Count word frequency using dictionary.


```python
inp_str=input("Enter the sentence to count word frequency : ")
inp_list=inp_str.split()
frequency={}
for word in inp_list:
  frequency[word]=frequency.get(word,0)+1
print(f"------Word Frequency-----------\n{frequency}")
```

###3.Sort dictionary by values.


```python
dict1={"A":1000,"B":299,"C":1990,"D":199,"E":99}
print(f"-------------sort dictionary by values-------\nInput Dictionary :\n{dict1}")
sorted_dict=dict(sorted(dict1.items(),key=lambda item:item[1],reverse=True))
print(f"Sorted Dictionary :\n{sorted_dict}")
```

###4.Find employee with highest salary.



```python
emp_dict={
    "emp_id":[101,102,104,112,145,199],
    "salary":[1000,2000,4590,9890,900,2000]
}
print(f"---------------Employee with Highest Salary-------------\nInput Dictionary :\n{emp_dict}")
salaries=emp_dict["salary"]
high_salary=max(salaries)
index=salaries.index(high_salary)
ids=emp_dict["emp_id"][index]
print(f"Employee with Highest Salary : {ids} with Salary : Rs.{high_salary}")

sal=max(emp_dict,key=emp_dict.get)
print(f"Employee with Highest Salary : {sal} with salary : {emp_dict[sal]}")
```

##Advanced


###1.Analyze product sales dataset:
{
 "Laptop": 10,
 "Mouse": 50,
 "Keyboard": 30
}
Find:
highest sold product
total items sold
percentage contribution



```python
pdt={
    "Laptop":10,"Mouse":20,"Keyboard":30,"Printer":25
}
most_sold_pdt=max(pdt,key=pdt.get)
qty=pdt[max(pdt,key=pdt.get)]
sum_qty=sum(pdt.values())
print(f"----------------Analyse Product Sales-------------\nDataset : {pdt}\n")
print(f"1.Highest Sold Product = {most_sold_pdt}")
print(f"2.Highest Sold Product Quantity = {qty}")
print(f"3.Highest Sold % Contribution = {round((qty/sum_qty)*100)}%")
```
