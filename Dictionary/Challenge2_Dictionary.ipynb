## Dictionary-Based Questions

### Create a dictionary from two lists

```python
list1=["A","B","C","D"]
list2=[1000,1100,2000,999]
emp_dict={key:value for key,value in zip(list1,list2)}
emp=dict(zip(list1,list2))
print(emp_dict)
print(emp)
```

### Count frequency of elements

```python
inp_str=input("Enter the sentence to count the frequency of elements : ")
print(f"Original Sentence : {inp_str}")
frq={}
inp_list=inp_str.split()
for word in inp_list:
  frq[word]=frq.get(word,0)+1
print(f"Frequency of Words : \n{frq}")
```

### Merge two dictionaries

```python
a={'A': 1000, 'B': 1100, 'C': 2000, 'D': 999}
b={'I': 3, 'love': 3, 'to': 3, 'eat': 1, 'cook': 1, 'dance': 1, 'and': 1, 'sing': 1}
print(f"Merge 2 dictionaries a and b:\na = {a}\nb = {b}")
print(f"Merged Dictionary = {a|b}")
```

### Find key with maximum value

```python
a={'A': 1000, 'B': 1100, 'C': 2000, 'D': 999}
print(f"input dict = {a}\nMaximum Value's Key = {max(a,key=a.get)}")
```

### Sort dictionary by value

```python
a={'A': 1000, 'B': 1100, 'C': 2000, 'D': 999}
sorted_dict=dict(sorted(a.items(),key=lambda item:item[1],reverse=True))
print(f"Input dict = {a}\nSorted Dictionary by value = {sorted_dict}")
```

### Check if key exists

```python
a={'A': 1000, 'B': 1100, 'C': 2000, 'D': 999}
print(f"Input dict = {a}")
print(f"Input dict after checking key 'E' exists  = {a.get('E',"No Key Found")}")
```

### Nested dictionary access

```python
customer={
    "Saranya": {
        "fname" : "Saranya","lname":"Thirunavukkarasu"
    },
    "Mithra" : {
        "fname" : "Mithra","lname":"Thirunavukkarasu"
    }
}
print(f"---------Nested Dictionary Access---------\nInput Dictionary = {customer}\nAll Keys = {customer.keys()}")
print(f"All Keys = {customer.keys()}\nAll Values = {customer.values()}\n\nDetailed View:")
for customer, details in customer.items():

    print(f"\nCustomer : {customer}")

    for key, value in details.items():

        print(f"{key} : {value}")
```

### Group data using dictionary

```python
employees = [
    ("Saran", "IT"),
    ("Arun", "HR"),
    ("Vijay", "IT"),
    ("Kumar", "Finance"),
    ("Rahul", "HR")
]
grouped_data = {}
for name,dept in employees:
  if dept not in grouped_data:
     grouped_data[dept] = []
  grouped_data[dept].append(name)
print(f"Grouped Data : {grouped_data}")
```

### Invert a dictionary

```python
student_marks = {
    "Saran": 85,
    "Arun": 90,
    "Vijay": 78
}
print(f"Input Dictionary = {student_marks}")
inverted_dict={value:key for key,value in student_marks.items()}
print(f"Inverted Dictionary = {inverted_dict}")
```

### Access the value of a given key with get()

```python
a={'A': 1000, 'B': 1100, 'C': 2000, 'D': 999}
print(f"Input dict = {a}")
print(f"Input dict after get('A') = {a.get('A')}")
```

### Add a new key-value pair to a dictionary

```python
a={'A': 1000, 'B': 1100, 'C': 2000, 'D': 999}
print(f"Input dict = {a}")
a['E']=799
print(f"Input dict after adding 'E':799 = {a}")
```

### Update an existing value in a dictionary

```python
a={'A': 1000, 'B': 1100, 'C': 2000, 'D': 999}
print(f"Input dict = {a}")
a['A']=888
print(f"Input dict after update('A'):888 = {a}")
```

### Delete a key from a dictionary

```python
a={'A': 1000, 'B': 1100, 'C': 2000, 'D': 999}
print(f"Input dict = {a}")
a.pop('A')
print(f"Input dict after pop('A') = {a}")
```
