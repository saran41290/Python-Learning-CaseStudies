Instructions:

Take a string input from the user using input().

• isdigit() for numbers

```python
#Take a string input from the user using input().
#Convert the text into lowercase using lower().
inp_str=input("Enter the string : ").lower()
vowels=cons=digits=specialchar=0
#Use a for loop to iterate through each character in the string.
#Count: • Total vowels • Total consonants • Total digits • Total special characters
for char in inp_str:
  if char.isalpha():
    if char in 'aeiou':
      vowels+=1
    else:
      cons+=1
  elif char.isdigit():
      digits+=1
  else:
      specialchar+=1
#Display all counts using print()
print(f"Input String : {inp_str}\nTotal Vowels = {vowels}\nTotal Consonants = {cons}\nTotal Digits = {digits}\nTotal SpecialCharacters = {specialchar}")
```

## Restricted Word Detection System

Create a list of words containing normal and restricted words.

```python
#Create a list of words containing normal and restricted words.
words_list=['Good','Bad','Shit','Shut','Happy','Close','Normal','Delete','Bad','Ugly','Del']
restricted_list=['Delete','Del']
safe_words=[]
restricted_count=0
#Use a for loop to read words one by one.
#Check whether the current word is a restricted word using if condition.
#If the restricted word is found: • Display "Restricted word detected" • Stop the loop using break
for word in words_list:
  if word in restricted_list:
    restricted_count+=1
    print(f"Restricted word detected : '{word}'")
  else :
    safe_words.append(word)
#Otherwise display all safe words using print()
print(f"All Safe Words : {safe_words}\nTotal Restricted words = {restricted_count}")
```

```python

#Create a function called marks_analysis() and pass the marks list to the function
#Return total and average marks
def marks_analysis(inp_marks):
  total=sum(inp_marks)
  average=total/len(inp_marks)
  return total,average


#Create a list called marks
mark_list=[99,80,79,69,98,100,79,65,45]

#call the function
total,average=marks_analysis(mark_list)

#display: Student Marks, Total Marks, Average Marks
print(f"-------------Student Marks Analysis using Function--------------------------\nInput Marks = {mark_list}")
print(f"Total marks = {total}\nAverage marks = {average:.0f}")
```
